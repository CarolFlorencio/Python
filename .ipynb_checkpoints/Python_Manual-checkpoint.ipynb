```python
#Exemplo de algoritmo para verificar o percentual do terreno ocupado pela garagem
Largura_Garagem=float(input("Entre com a largura da garagem em metros: "))
Profundidade_Garagem=float(input("Entre com a profundidade da garagem em metros: "))
#Abaixo, cálculo da área da garagem
Area_Garagem=Largura_Garagem * Profundidade_Garagem
Largura_Terreno=float(input("Entre com a largura do  terreno em metros: "))
Profundidade_Terreno=float(input("Entre com a pro fundidade do terreno em metros: "))
# Abaixo, cálculo da área do terreno
Area_Terreno=Largura_Terreno * Profundidade_Terreno
# Agora, cálculo do percentual de ocupação
Percentual = ((Area_Garagem)/(Area_Terreno))* 100
print("Percentual de ocupação:",Percentual)
```

```python
#Algoritmo para descobrir o preço de custo e preço final para os produtos de uma fabrica de sapatos.
#Ler valor de custo do couro
custo_couro = float(input("Insira o valor de custo do couro em reais(R$):"))
#Ler valor de custo do solado
custo_solado = float(input("Insira o valor de custo do solado em reais(R$):"))
#Ler valor de custo dos cordoes e ilhoses
custo_cordoes_ilhoses = float(input("Insira o valor de custo dos cordões e ilhoses em reais(R$):"))
#Ler valor dos insumos
insumos = float(input("Insira o valor dos insumos em reais(R$):"))
#Ler valor de custo de mão de obra
mao_de_obra = float(input("Ler valor de custo de mão de obra em reais(R$):"))
#Ler valor de custo de Marketing
custo_marketing = float(input("Ler valor de custo de marketing em reais(R$):"))
#Ler valor de custo de vendas
custo_vendas = float(input("Ler valor de custo de vendas em reais(R$):"))
#Calculando preço de custo
preco_custo = custo_couro*0.3+custo_solado*0.2+custo_cordoes_ilhoses*0.05+insumos*0.05+mao_de_obra*0.2+custo_marketing*0.1+custo_vendas*0.1
#Calculando preço de lucro do fabricante
preco_lucro = preco_custo*1.3
#Calculando preço com perdas de capital
preco_perda = preco_lucro*1.15
#Calcular preço com impostos federais IPI e COFINS
preco_IPI_COFINS = preco_perda*1.25
#Calculando preço com margem de vendas
preco_margem_vendas = preco_IPI_COFINS*1.25
#Calculando preço com impostos estaduais e municipais
preco_impostos = preco_margem_vendas*1.30
#Definindo preço final
preco_final = preco_impostos
#Exibindo preço de custo e preço final calculados
print("O preço de custo será R$",preco_custo)
print("O preço final será R$",preco_final)
```

```python
#DEFINIÇÃO DA FUNÇÃO COM RETORNO 
def getAnswer(answerNumber):
    if answerNumber == 1:
           return 'It is certain'
    elif answerNumber == 2:
        return 'It is decidedly so'
    elif answerNumber == 3:
        return 'Yes'
    elif answerNumber == 4:
        return 'Reply hazy try again'
    elif answerNumber == 5:
        return 'Ask again later'
    elif answerNumber == 6:
        return 'Concentrate and ask again'
    elif answerNumber == 7:
        return 'My reply is no'
    elif answerNumber == 8:
        return 'Outlook not so good'
    elif answerNumber == 9:
        return 'Very doubtful'
r = ri(1, 9)
fortune = getAnswer(r)
print(fortune)
```

```python
#EXEMPLO DE USO DE CLASSE DE DA FUNÇÃO STR E EVAL
class Ponto:
	def __init__ (self, x, y):
		self.x = x
		self.y = y
	def __str__(self):
		return "({}, {})".format(self.x, self.y)
	def repr(self):
		return "Ponto({}, {})".format(self.x + 1, self.y + 1)
p1 = Ponto(1, 2)
p2 = eval(p1.repr())
print(p1)
print(p2)
```

```python
#Criação de uma função para calcular o IMC e seu uso
def IMC():
    massa = float(input("Digite a sua massa em KG:"))
    altura = float(input("Digite a sua altura em cm:"))
    imc = massa/(altura**2)
    print( "Seu IMC é: %2.2f" % imc) 
IMC()
```

```python
#Criação de uma função que é uma pequena calculadora
def calculadora (): 
	num1 = float(input("Digite um número:"))
	num2 = float(input("Digite outro número:"))
	operacao = input("Digite a operação(+, -, *, /):")
	
	resultado = num1
	if operacao == '+':
		resultado += num2
	elif operacao == '-':
		resultado -= num2
	elif operacao == '*':
		resultado *= num2
	elif operacao == '/':
		resultado /= num2
	else:
		print("Operação não reconhecida. Tente de novo!")
	print("O resultado é: %f" % resultado)
calculadora()
```

```python
#Exemplo de função que calcula o novo salário em uma empresa 
salarioMinimo = 1000
def calculaNovoSalario (salarioBruto, descontoPlanoSaude, descontosDiversos):
    salarioLiquido = salarioBruto
    salarioLiquido -= descontosDiversos
    salarioLiquido -=descontoPlanoSaude * (50/100)
    print("Seu antigo salário era dê R$",salarioLiquido)
    salarioLiquido +=salarioMinimo * (8/100)
    salarioLiquido += salarioBruto * (5/100)
    return salarioLiquido
salarioB = float(input("Digite o seu salário bruto:"))
descontoPS = float(input("Digite o desconto do Plano de Saúde:"))
descontoD = float (input("Digite os seus descontos diversos:"))
salarioL = calculaNovoSalario(salarioB, descontoPS, descontoD)
print("Seu novo salário líquido será de R$",salarioL)
```

# Utilizando o método `strtime()`

```python
#Exemplo strftime
from datetime import datetime

hoje = datetime.now() #Lê a data e tempo atual

#Imprime a informações separadas
ano = hoje.strftime("%Y")
print("Ano:",ano)

mes = hoje.strftime("%m")
print("Mês:",mes)

dia = hoje.strftime("%d")
print("Dia:",dia)
```

```python
def cadastro_cliente():
    nome = input('Digite o nome do cliente: ')
    idade = int(input('Digite a idade do cliente: '))
    cpf = input('Digite o cpf do cliente: ')
    return [nome,idade,cpf]
def exibir_clientes(tabelaClientes):
    print('\n\n')
    cabecalho = ['Nome', 'Idade', 'CPF']
    print(*cabecalho, sep='\t')
    for cliente in tabelaClientes:
        print(*cliente, sep='\t')
def salvar_arquivo(tabelaClientes):
    meu_arquivo = open("clientes.txt", "w", encoding="utf-8")
    cabecalho = ['Nome', 'Idade', 'CPF']
    print(*cabecalho, sep='\t',file=meu_arquivo)
    for cliente in tabelaClientes:
        print(*cliente, sep='\t', file=meu_arquivo)  
    meu_arquivo.close()
    print('\n\nO arquivo com os dados foi gerado com sucesso e está pronto para ser importado pelo novo sistema da empresa.')

qtdCliente = int(input('Digite a quantidade de clientes que deseja cadastrar: '))
tabCliente = []
for i in range(0,qtdCliente):
    tabCliente.append(cadastro_cliente())
exibir_clientes(tabCliente)
salvar_arquivo(tabCliente)
```
